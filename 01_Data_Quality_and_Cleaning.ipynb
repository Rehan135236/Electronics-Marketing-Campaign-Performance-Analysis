```python
outlier_columns = [
    'Impressions',
    'Reach',
    'Clicks',
    'Spend_USD',
    'Conversions',
    'Revenue_USD'
]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n===== {col} =====")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("Lower:", lower)
    print("Upper:", upper)
    print("Outlier count:", len(outliers))
```

```python
df['CTR'] = df['Clicks'] / df['Impressions']

df['CPC'] = df['Spend_USD'] / df['Clicks']

df['CPM'] = (df['Spend_USD'] / df['Impressions']) * 1000

df['CPA'] = df['Spend_USD'] / df['Conversions']

df['ROAS'] = df['Revenue_USD'] / df['Spend_USD']

df['Conversion_Rate'] = df['Conversions'] / df['Clicks']

df['Purchase_Rate'] = df['Purchases'] / df['Add_to_Cart']
```

```python
import numpy as np

df['CTR'] = np.where(
    df['Impressions'] > 0,
    df['Clicks'] / df['Impressions'],
    np.nan
)

df['CPC'] = np.where(
    df['Clicks'] > 0,
    df['Spend_USD'] / df['Clicks'],
    np.nan
)

df['CPM'] = np.where(
    df['Impressions'] > 0,
    (df['Spend_USD'] / df['Impressions']) * 1000,
    np.nan
)

df['CPA'] = np.where(
    df['Conversions'] > 0,
    df['Spend_USD'] / df['Conversions'],
    np.nan
)

df['ROAS'] = np.where(
    df['Spend_USD'] > 0,
    df['Revenue_USD'] / df['Spend_USD'],
    np.nan
)

df['Conversion_Rate'] = np.where(
    df['Clicks'] > 0,
    df['Conversions'] / df['Clicks'],
    np.nan
)

df['Purchase_Rate'] = np.where(
    df['Add_to_Cart'] > 0,
    df['Purchases'] / df['Add_to_Cart'],
    np.nan
)
```

```python
platform_budget = df.groupby('Platform').agg(
    Spend=('Spend_USD', 'sum'),
    Revenue=('Revenue_USD', 'sum'),
    Conversions=('Conversions', 'sum'),
    Clicks=('Clicks', 'sum'),
    Impressions=('Impressions', 'sum')
).reset_index()
```

```python
platform_budget['ROAS'] = (
    platform_budget['Revenue'] /
    platform_budget['Spend']
)

platform_budget['CPA'] = (
    platform_budget['Spend'] /
    platform_budget['Conversions']
)

platform_budget['CTR'] = (
    platform_budget['Clicks'] /
    platform_budget['Impressions']
)

platform_budget['Spend_Share'] = (
    platform_budget['Spend'] /
    platform_budget['Spend'].sum()
)

platform_budget['Revenue_Share'] = (
    platform_budget['Revenue'] /
    platform_budget['Revenue'].sum()
)
```

```python
platform_budget.sort_values(
    'ROAS',
    ascending=False
).round(3)
```

```python
platform_country = df.groupby(
    ['Platform', 'Country']
).agg(
    Spend=('Spend_USD', 'sum'),
    Revenue=('Revenue_USD', 'sum'),
    Conversions=('Conversions', 'sum'),
    Clicks=('Clicks', 'sum'),
    Impressions=('Impressions', 'sum')
).reset_index()
```

```python
platform_country['ROAS'] = (
    platform_country['Revenue'] /
    platform_country['Spend']
)

platform_country['CPA'] = (
    platform_country['Spend'] /
    platform_country['Conversions']
)

platform_country['CTR'] = (
    platform_country['Clicks'] /
    platform_country['Impressions']
)

platform_country['Spend_Share'] = (
    platform_country['Spend'] /
    platform_country.groupby('Platform')['Spend'].transform('sum')
)

platform_country['Revenue_Share'] = (
    platform_country['Revenue'] /
    platform_country.groupby('Platform')['Revenue'].transform('sum')
)
```

```python
platform_country.sort_values(
    'ROAS',
    ascending=False
).round(3)
```

```python
total_spend = df['Spend_USD'].sum()
total_revenue = df['Revenue_USD'].sum()
total_conversions = df['Conversions'].sum()
total_purchases = df['Purchases'].sum()
total_clicks = df['Clicks'].sum()
total_impressions = df['Impressions'].sum()

overall_roas = total_revenue / total_spend
overall_ctr = total_clicks / total_impressions
overall_conversion_rate = total_conversions / total_clicks

print("===== EXECUTIVE KPIs =====")
print(f"Total Spend: ${total_spend:,.2f}")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Conversions: {total_conversions:,}")
print(f"Total Purchases: {total_purchases:,}")
print(f"Overall ROAS: {overall_roas:.2f}")
print(f"Overall CTR: {overall_ctr:.2%}")
print(f"Overall Conversion Rate: {overall_conversion_rate:.2%}")
```
